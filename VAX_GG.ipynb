```python
soup = BeautifulSoup(html_string)
```

```python
print(soup.prettify())
```

```python
#tag_object_model=soup.find_all('span',{'class':'product-tile-model text-graydark text-nowrap'})
tag_object_model=soup.find_all('div',{'class':'product-tile-main clearfix'})
tag_object_model
```

```python
model_no_list = []
for product in tag_object_model:
    model = product.find('span',{'class':'product-tile-model text-graydark text-nowrap'})
    model_no = model.get_text('', strip = True)
    model_no_list.append(model_no)
model_no_list
```

```python
#https://stackoverflow.com/questions/14264102/python-beautifulsoup-typeerror-sequence-item-0-expected-string-tag-found
#https://docs.python.org/3/tutorial/errors.html#handling-exceptions

model_price_list = []
for product in tag_object_model:
    try:
        price = product.find('span',{'class':'pricepoint-price'})
        RRP = price.get_text('', strip = True)
    except:
        RRP = '0'
    model_price_list.append(RRP)   
model_price_list
```

```python
df = pd.DataFrame(model_no_list, columns = ['Material_No'])
df
```
