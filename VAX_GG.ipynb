```python
import numpy as np
import pandas as pd
```

```python
with open('Vax - Vacuums & Cleaners - The Good Guys.html', 'r') as f:
    html_string = f.read()
html_string
```

```python
soup = BeautifulSoup(html_string)
```

```python
print(soup.prettify())
```

```python
tag_object_model=soup.find_all('span',{'class':'product-tile-model text-graydark text-nowrap'})
tag_object_model
```

```python
model_no_list = []
for model in tag_object_model:
    model_No = model.contents
    model_no_list.append(model_No)
model_no_list
```

```python
tag_object_price=soup.find_all('span',{'class':'pricepoint-price'})
tag_object_price
```

```python
# np.dtype(tag_object_price)
```

```python
#https://stackoverflow.com/questions/14264102/python-beautifulsoup-typeerror-sequence-item-0-expected-string-tag-found

price_list = []
for price in tag_object_price:
    RRP = price.get_text('', strip = True)
    price_list.append(RRP)
price_list
```

```python
df = pd.DataFrame(model_no_list, columns = ['Material_No'])
df
```

```python
df['RRP'] = pd.Series(price_list)
df
```
